### A hands-on taste of the biology: simulating cell growth

The video mentioned that cancer's core problem is uncontrolled division. Let's make that
concrete with a simple simulation. Healthy tissue keeps its cell count roughly steady (old
cells die off at about the same rate new ones are made). Cancer cells, on the other hand,
just keep doubling on a regular schedule — a pattern called **exponential growth**.

```python
import numpy as np
import matplotlib.pyplot as plt

days = np.arange(0, 30)  # simulate 30 days, day 0 to day 29

# Healthy tissue: count stays roughly flat (regulated turnover), with tiny random wobble
rng = np.random.default_rng(3)
healthy_cells = 1000 + rng.normal(0, 15, len(days))

# Cancer cells: start at just 10 cells, but DOUBLE every 3 days (exponential growth)
doubling_time = 3
cancer_cells = 10 * (2 ** (days / doubling_time))

plt.figure(figsize=(7,5))
plt.plot(days, healthy_cells, label='Healthy tissue (regulated)', color='steelblue')
plt.plot(days, cancer_cells, label='Cancer cells (exponential growth)', color='crimson')
plt.xlabel('Day')
plt.ylabel('Cell count')
plt.title('Regulated vs. uncontrolled cell growth')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

**Try it:** change `doubling_time` to a bigger number (like 6) or smaller number (like
1.5) and re-run. A shorter doubling time means a more aggressive, fast-growing cancer.

**Reflection:** notice how flat the cancer curve looks for the first ~15 days,
then it suddenly shoots upward. This is a real property of exponential growth — it can look
deceptively slow at first ("it's just a small lump") before becoming dramatic very quickly.
This is part of why catching disease early, before that steep part of the curve, matters so
much — and it's a big part of why medical imaging (which can catch a tumor before it's large
enough to feel or see any other way) is so valuable.

## 2. Chemistry: how does a contrast agent actually work?

You learned on Day 2 that MRI signal comes from hydrogen atoms in water. Sometimes,
radiologists inject a **contrast agent** into a patient's bloodstream right before a scan to
make certain tissues (like tumors or blood vessels) show up more clearly. The most common
MRI contrast agents are built around a metal called **gadolinium** — and the chemistry
behind make it safe to inject.

**The core idea, in a few steps:**
1. Gadolinium is *paramagnetic* — it's strongly affected by magnetic fields, which is exactly
   why it's useful in MRI. It changes how quickly nearby water molecules "relax" after being
   disturbed by the machine's radio pulse (remember T1/T2 from Day 2?), making the surrounding
   tissue appear brighter.
2. On their own, free gadolinium ions (Gd³⁺) are toxic to the body — they can interfere with
   processes that normally use calcium ions, which are a similar size and charge.
3. So chemists attach the gadolinium to a **chelating agent** — a molecule that wraps around
   the metal ion from multiple points at once, like a claw (the word "chelate" comes from the
   Greek word for claw). This traps the gadolinium in a stable "cage," so it passes through
   the body and gets safely filtered out by the kidneys without ever really contacting
   tissue directly.

### A hands-on taste of the chemistry: simulating clearance from the body

Chelated contrast agents are designed to be cleared from the bloodstream quickly and safely,
mostly by the kidneys. This kind of clearance follows a pattern you may have seen in AP
Chemistry: **first-order kinetics** — the concentration decreases by the same *fraction*
every fixed time period, which produces an exponential decay curve with a **half-life** (the
time it takes for half of the remaining amount to clear).

Let's simulate this for a real gadolinium contrast agent, which typically has a blood
half-life of roughly 1.5 hours in a patient with normal kidney function.

```python
import numpy as np
import matplotlib.pyplot as plt

hours = np.linspace(0, 12, 100)  # simulate the first 12 hours after injection

half_life_hours = 1.5   # typical for a gadolinium-based contrast agent
initial_concentration = 100   # arbitrary starting units, right after injection

# The first-order decay formula: C(t) = C0 * (1/2)^(t / half_life)
concentration = initial_concentration * (0.5) ** (hours / half_life_hours)

plt.figure(figsize=(7,5))
plt.plot(hours, concentration, color='darkgreen')
plt.axhline(50, color='gray', linestyle='--', linewidth=1)
plt.axvline(half_life_hours, color='gray', linestyle='--', linewidth=1)
plt.xlabel('Hours since injection')
plt.ylabel('Contrast agent concentration in blood (%)')
plt.title('Contrast agent clearance over time')
plt.grid(alpha=0.3)
plt.show()

print(f"After one half-life ({half_life_hours} hours), concentration drops to 50%.")
print(f"After 5 half-lives (~{5*half_life_hours} hours), less than 5% remains.")
```

**Try it:** change `half_life_hours` to simulate a patient with reduced kidney function,
where clearance might take much longer (try `6` or `10`). What does slower clearance mean for
how long gadolinium stays in the body, and why might that matter for patient safety? (This is
part of why doctors check kidney function before giving contrast agents.)

## 3. Physics: how much radiation is actually in a CT scan, and is it safe?

You learned on Day 3 that CT uses X-rays, and MRI/ultrasound don't. That naturally raises a
question: how much radiation is actually involved, and how do hospitals decide it's safe?

📺 **Watch:** [Measuring Radiation Exposure: What is a Sievert? — short explainer](https://www.youtube.com/watch?v=vRYwUPBZ-J4)

**Some concrete numbers to anchor the discussion:**
- The average person receives about 3 millisieverts (mSv) per year just from natural
  background radiation (cosmic rays, radon gas, etc.)
- A chest CT scan is roughly 6-7 mSv — a bit more than two years' worth of background
  radiation, delivered in one scan.
- Hospitals follow a principle called **ALARA** — "As Low As Reasonably Achievable." It
  doesn't mean avoiding radiation entirely; it means never using more than needed to get a
  useful image, and adjusting the dose based on patient size, age, and what's being looked
  for.

**Reflection:**
- Why might a pediatric (children's) CT protocol use a different, lower dose than an adult
  protocol scanning the same body part?
- Given that MRI and ultrasound don't use ionizing radiation at all, why do you think CT is
  still used so often instead of always defaulting to one of those?

### A hands-on taste of the physics: comparing radiation doses

Numbers like "6.8 mSv" don't mean much on their own — they're most useful compared side by
side. Let's build a simple bar chart comparing some common radiation exposures (these are
commonly cited rounded approximate values, since actual doses vary by patient and equipment).

```python
import matplotlib.pyplot as plt

# Approximate effective doses, in millisieverts (mSv) - rounded, for comparison purposes
sources = [
    "Cross-country\nflight",
    "Chest\nX-ray",
    "Background\nradiation (1 year)",
    "Mammogram",
    "Chest\nCT scan",
    "Abdomen/pelvis\nCT scan",
]
doses_mSv = [0.03, 0.1, 3, 0.4, 7, 10]

plt.figure(figsize=(8,5))
bars = plt.bar(sources, doses_mSv, color='steelblue')
plt.ylabel('Approximate effective dose (mSv)')
plt.title('Comparing radiation doses from different sources')
plt.xticks(rotation=20, ha='right')

# Label each bar with its value
for bar, dose in zip(bars, doses_mSv):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.1, str(dose),
              ha='center', va='bottom', fontsize=9)

plt.tight_layout()
plt.show()
```

**Try it:** add one more bar to the `sources` and `doses_mSv` lists for a source you're
curious about (a quick web search for "radiation dose chart" will turn up plenty of
comparisons) and re-run the cell.

**Reflection:** notice how a chest CT scan is roughly 20x higher dose than a
chest X-ray, but still much lower than doses known to cause immediate harm (acute radiation
sickness starts around 1000 mSv, over 100x a chest CT). This kind of visual, relative
comparison is exactly the kind of reasoning behind the ALARA principle — weighing a real but
small risk against real diagnostic benefit.

## 4. Genomics: from DNA to "radiogenomics" — combining imaging with genetics

Zoom out from cells (biology) and atoms (chemistry, physics) to the *code* that builds a
person in the first place: their genome.

📺 **Watch:** [Richard Resnick: Welcome to the Genomic Revolution — TED Talk, ~10 min](https://www.youtube.com/watch?v=u8bsCiq6hvM)

(This talk is from 2011, so some of the cost/speed numbers for DNA sequencing are now
outdated — sequencing has gotten even cheaper and faster since then — but the core ideas
about what genomics makes possible still hold up well.)

**Now, the connection to imaging:** researchers have discovered that certain patterns you
can see or measure in a medical image are statistically correlated with what's happening in
a patient's genes — for example, particular texture or shape features in a tumor on an MRI
scan often correlate with specific genetic mutations driving that tumor's growth. This
combined field is called **radiogenomics** ("radio" for radiology + "genomics"), and it's an active area of research.

**Why does this matter?** A biopsy (physically removing tissue to test its genetics) is
invasive, can only sample one small part of a tumor, and can't easily be repeated often. An
MRI or CT scan is non-invasive and can capture information about the *entire* tumor at once.
If researchers can reliably predict genetic information from imaging alone, that could mean
faster, safer, and more complete information for doctors — without an invasive procedure
every time.

### A hands-on taste of genomics: reading DNA in Python

Before we get to a radiogenomics demo, let's get hands-on with genomic data itself. DNA is a
sequence made of just four letters: A, C, G, and T (representing four chemical bases:
adenine, cytosine, guanine, thymine). In genomics, a huge amount of analysis starts with
simple operations on these letter sequences.

```python
# A short, made-up DNA sequence (in reality these are millions of letters long!)
dna_sequence = "ATGGCCATTGTAATGGGCCGCTGAAAGGGTGCCCGATAG"

print("Sequence:", dna_sequence)
print("Length:", len(dna_sequence), "bases")
```

```python
# Count how many times each base appears using .count()
count_a = dna_sequence.count("A")
count_t = dna_sequence.count("T")
count_g = dna_sequence.count("G")
count_c = dna_sequence.count("C")

print("A:", count_a)
print("T:", count_t)
print("G:", count_g)
print("C:", count_c)
```

```python
# GC content (the percentage of bases that are G or C) is a real, commonly used genomic
# measurement - it varies between species and even between regions of the same genome,
# and affects how stable a piece of DNA is.
gc_content = 100 * (count_g + count_c) / len(dna_sequence)

print(f"GC content: {gc_content:.1f}%")
```

**Try it:** change `dna_sequence` at the top to a different string of A, C, G, T
letters (any length, any order) and re-run all three cells. Watch how the counts and GC
content change.

**Reflection:** genomics researchers compute GC content and much more complex
statistics like this across entire genomes (billions of letters). This tiny example is the
same basic idea as real genomic analysis, just at a much smaller scale — a lot of genomics
really is counting and pattern-finding in long strings of letters, just with far more data and
far more sophisticated methods.

### Now, a small hands-on taste of radiogenomics specifically

Real radiogenomics research involves large datasets, careful statistics, and often the same
kind of AI you'll build starting tomorrow. But the core idea — checking whether an imaging
measurement and a genetic measurement are related — can be demonstrated with a small
simulated example.

Below, we'll invent data for 40 simulated patients: for each one, a simple "imaging feature"
(how heterogeneous / textured their tumor looks on a scan) and a simulated "genomic marker"
(the activity level of one specific gene in their tumor). Then we'll check whether the two
are related.

```python
import numpy as np
import matplotlib.pyplot as plt

# Make this example reproducible - using the same seed always gives the same "random" data
rng = np.random.default_rng(7)

n_patients = 40

# Simulate a genomic marker (e.g., activity level of a specific gene), roughly 0-10
genomic_marker = rng.uniform(0, 10, n_patients)

# Simulate an imaging feature that is REALISTICALLY CORRELATED with the genomic marker,
# plus some random noise (real biology is never a perfectly clean relationship)
imaging_feature = 2.0 * genomic_marker + rng.normal(0, 4, n_patients)

print("First 5 simulated patients:")
for i in range(5):
    print(f"  Patient {i+1}: genomic marker = {genomic_marker[i]:.1f}, imaging feature = {imaging_feature[i]:.1f}")
```

```python
# Plot the two measurements against each other - this is called a scatter plot,
# and it's one of the most common first steps in any radiogenomics analysis
plt.figure(figsize=(6,6))
plt.scatter(genomic_marker, imaging_feature, color='darkorange', edgecolor='black')
plt.xlabel('Genomic marker (simulated gene activity)')
plt.ylabel('Imaging feature (simulated tumor texture score)')
plt.title('Does the imaging feature relate to the genomic marker?')
plt.grid(alpha=0.3)
plt.show()
```

```python
# np.corrcoef computes the correlation coefficient - a number between -1 and 1 that
# measures how strongly two things are related. Close to 1 (or -1) means strongly related;
# close to 0 means not related at all.
correlation_matrix = np.corrcoef(genomic_marker, imaging_feature)
correlation = correlation_matrix[0, 1]

print(f"Correlation between imaging feature and genomic marker: {correlation:.2f}")
```

**Reflection:** a correlation doesn't automatically mean one thing
*causes* the other — that's a famous rule in statistics ("correlation isn't causation").
What are some other explanations for why an imaging feature and a genomic marker might be
correlated, besides one directly causing the other?

**Try it:** go back to the code cell that creates `imaging_feature` and change the `2.0` to
`0.2`, then re-run all three cells. What happens to the scatter plot and the correlation
number? This simulates a much weaker real-world relationship — which is often what
researchers actually find, and why radiogenomics studies need large numbers of patients to
detect real but subtle patterns reliably.
